# 1. TypedDict

```python
# Normal Dictionary
d = {"name": "Naruto", "age": 13} 
d
```

### Pros 
- allows efficient data retrieval
- flexiably and easy implementation

### Cons
- dont check wheather data is in particular
    -  structure
    - type

this can be source of a lot of logical errors <br>
Solution: <b>Typed dictionary</b>    

```python
from typing import TypedDict
```

```python
class Movie:
    name: str 
    age: int
```

```python
movie = Movie(name="Naruto", age=10)
```

```python
# You should inherit that
class Movie(TypedDict):
    name: str 
    age: int
```

```python
movie = Movie(name="Naruto", age=10)
```

```python
movie
```

```python
# Pylance shows warning
movie_1 = Movie(name="Batman", age="30")
```

```python
# !! no warnings here
movie_1
```

```python
movie["age"]
```

```python
#  can't use as an object attribute
movie.age
```

# 2. Union

Union says that the value can be in more than one type

```python
from typing import Union
```

```python
def square(a: Union[int, float]):
    return a**2
```

```python
square(3)
```

```python
square(4.6)
```

```python
# also type checking warning
square("89")
```

# 3. Optional

```python
from typing import Optional
```

```python
def nice_message(name: Optional[str]):
    if name: 
       print(f"Hi there, {name}")
    else:
        print("Hi stranger.")    
```

```python
nice_message("Naruto")
```

```python
nice_message(3)
```

```python
# I have to pass None Explisitly
nice_message()
```

```python
nice_message(None)
```

# 4. Any

```python
from typing import Any
```

```python
def print_value(x: Any):
    print(x)
```

```python
print_value(4)
```

```python
print_value()
```

```python
print_value(None)
```
